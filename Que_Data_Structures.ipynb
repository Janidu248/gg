```python
class Node:

  def __init__(self,data):
    self.data = data
    self.next = None
```

```python
class queue :
  def __init__(self):
    self.front = None
    self.rear = None


  def enqueue(self,data):
    Newnode  = Node(data)

    if self.rear is None:

     self.front = Newnode
     self.rear = Newnode
     return

    self.rear.next = Newnode
    self.rear = Newnode
```
